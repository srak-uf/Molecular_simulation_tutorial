## Lennard-Jones potential

In this section, the Lennard–Jones (LJ) potential is described. LJ potential is one of the most important potential both historically and practically. LJ potential was developed ~100 years ago (1924! Present 2024 is 100th anniversary!!!!). The function is simply written as follows: 

$$U_{ij}(r_{ij}) =  4\varepsilon_{ij}\left[\left(\frac{\sigma_{ij}}{r_{ij}}\right)^{12}-\left(\frac{\sigma_{ij}}{r_{ij}}\right)^{6}\right],$$

where $r_{ij}$ is the distance between particles $i$ and $j$, $\sigma_{ij}$ and $\epsilon_{ij}$ are the empirical parameters. Noted that the attractive long-range $r^{–6}$ behavior for the dispersion force was introduced in 1930 by London and has theoretical background. The LJ potential is even now widely used for describing non-bonded interaction in some popular organic-molecule force fields (GAFF, AMBER, CHARMM, OPLS, and so on).  
  
In Argon case, the empical parameter $\sigma_{ij}$ and $\epsilon_{ij}$ is given by as follows:

$$\sigma_{ij} = 3.40 \,\text{Å},$$
$$\epsilon_{ij} = 0.1034 \,\text{eV}.$$

The Lennard–Jones potential is already implemented in `ASE`. The potential property is examined in following codes.

```python
# Definition of Lennard-Jones potential, rc is the cutoff radius (in Angstrom)
sigma = 3.4 # Angstrom
epsilon = 0.01034 # eV
argon_lj = LennardJones(epsilon=epsilon, sigma=sigma, rc=100)
```

The `Atoms` object is a collection of atoms. Here is how to define a two Argons system:

```python
from ase import Atoms
d = 5# 1.5
ar_ar = Atoms("ArAr", positions=[(0, 0, 0), (d, 0, 0)])
# View the Ar-Ar configuration (from ase.visualize import view)
view(ar_ar)
```

To calculate energy by ASE,   
1. `Atoms.calculator` should be set. In this case, argon_lj is used.
2. `Atoms.get_potential_energy()` provide the potential energy

```python
ar_ar.calc = argon_lj
ar_ar.get_potential_energy()
```

```python
# Scan the potential energy as a function of the distance between the two atoms
distances = np.linspace(3, 10, 100)
energies = []
for d in distances:
    ar_ar.positions[1] = (d, 0, 0)
    energies.append(ar_ar.get_potential_energy())
```

```python
# Plot the potential energy as a function of the distance
plt.plot(distances, energies)
plt.xlabel("Distance (Angstrom)")
plt.ylabel("Potential energy (eV)")
```
