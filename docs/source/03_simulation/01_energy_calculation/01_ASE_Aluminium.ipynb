# ASEの基礎

## Goal of this tutorial
> The goal of this tutoal is to know the basic function of ASE.  
> - Atoms object
> - Calculator
> - Fileio  
> - Visualize


For MD simulation, the following information are necessary at least.  
1. Materials structure
2. Level of energy calculation  

The above two functions are implemented respectively as follows:
1. `ase.Atoms` class: Represent molecules and materials
2. `ase.calculators.calculator.Calculator` class: Calculate energy and force of the `Atoms` object  
    Calculator class can be created by yourself easily. Additionaly, ASE provides the wrapper of existing molecular simulation code as calculator &mdash; VASP, CP2K, Quantum Espresso, Gaussian, LAMMPS, GROMACS, and so on. 

In addition, the initial structure is often prepared by using the CIF file obtained by XRD measurment. The fileio function for any types of coordinate files (cif, xyz, pdb, POSCAR, and so on) is important.  
ASE also has good reader and writer functions.  
1. `ase.io.read`
2. `ase.io.write`

## EAM calculation of Aluminium

```python
from ase.build import bulk
from ase.calculators.eam import EAM

a = 4.05  # Angstrom lattice spacing
al = bulk('Al', 'fcc', a=a)
al
```

```python
# check the structure visually
from ase.visualize import view
view(al)
```

```python
# Replicate 5x5x5
al = al.repeat((2, 2, 2))
view(al)
```

```python
# Download of EAM potential file
# By adding !, you can execute terminal commands.
# wget is a command to download files
! wget https://www.ctcms.nist.gov/potentials/Download/1999--Mishin-Y-Farkas-D-Mehl-M-J-Papaconstantopoulos-D-A--Al/2/Al99.eam.alloy
```

```python
# Check the downloaded file
! ls
```

```python
# Atoms.calc = calculator
al.calc = EAM(potential='Al99.eam.alloy')
```

```python
print(f"Potential energy is {al.get_potential_energy()} eV")
```

```python
print("Al positions")
print(al.get_positions())

print("")

print("Al symbols")
print(al.get_chemical_symbols())
```

```python
print("Al forces")
print(al.get_forces())

print("")

print("Al masses")
print(al.get_masses())

print("")

print("Al center of mass")
print(al.get_center_of_mass())

print("")

print("Al2x2x2 volume")
print(f"{al.get_volume()} A^3")
```

## ASE default units
- Info: [https://wiki.fysik.dtu.dk/ase/ase/units.html](https://wiki.fysik.dtu.dk/ase/ase/units.html)  


> Physical units are defined in the ase/units.py module.  
> **Electron volts (eV), Ångström (Ang), the atomic mass unit and Kelvin are defined as 1.0.**  
> Other units are (amongst others) nm, Bohr, Hartree or Ha, kJ, kcal, mol, Rydberg or Ry, second, fs and kB. 

```python
from ase.units import Hartree, kJ, mol, m, kg
```

```python
# Easily convert units by using ASE
# 1 Hartree = 27.211... eV
print(1*Hartree)

# 1 eV = 3.674932...E-2 Hartree 
print(1 / Hartree)

# 1 eV = 96.485... kJ/mol
print(1 / kJ * mol)
```

## Fileio

```python
# save 5x5x5 structure
from ase.io import write
write('Al_2x2x2.xyz', al)
```

```python
# read the saved file
from ase.io import read
al_read = read('Al_2x2x2.xyz')
view(al_read)
```
